# DEMO KIỂM THỬ THỰC THI SUY LUẬN MÔ HÌNH ML (INFERENCE V1)

Notebook này minh họa quy trình load mô hình `student_score_model-1.joblib` và thực hiện suy luận (`predict`) trên dữ liệu học sinh với 11 đặc trưng chuẩn UCI Student Performance.

```python
import os
import json
import joblib
import pandas as pd
import numpy as np

print("Numpy version:", np.__version__)
print("Joblib version:", joblib.__version__)
```

## 1. Load Artifacts & Metadata

```python
model_path = os.path.abspath('../../ml/models/student_score_model-1.joblib')
metadata_path = os.path.abspath('../../ml/models/model_metadata.json')

with open(metadata_path, 'r', encoding='utf-8') as f:
    metadata = json.load(f)

model = joblib.load(model_path)
print("Loaded model successfully:", type(model))
print("Model metadata:", json.dumps(metadata, indent=2, ensure_ascii=False))
```

## 2. Chuẩn bị Dữ liệu Đầu vào (11 Features)

```python
sample_input = {
    'studytime': 3,     # 1-4
    'failures': 0,      # 0-4
    'absences': 2,      # 0-93
    'G1': 15,           # 0-20
    'school': 'GP',     # 'GP' or 'MS'
    'sex': 'F',         # 'F' or 'M'
    'age': 16,          # 15-22
    'internet': 'yes',  # 'yes' or 'no'
    'higher': 'yes',    # 'yes' or 'no'
    'goout': 2,         # 1-5
    'health': 4         # 1-5
}

df_input = pd.DataFrame([sample_input])
print("Input DataFrame:")
df_input
```

## 3. Thực thi model.predict()

```python
raw_prediction = model.predict(df_input)[0]
predicted_score = round(float(np.clip(raw_prediction, 0, 20)), 1)

print(f"=== KẾT QUẢ SUY LUẬN ===")
print(f"Điểm dự đoán (G3): {predicted_score} / 20")
print(f"Thuật toán: {metadata.get('model_name')}")
print(f"Phiên bản: v{metadata.get('version')}")
```
